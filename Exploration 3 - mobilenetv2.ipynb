# Exploration 3 - mobilenetv2

```python
import torch
from torchvision.models import mobilenet_v2, MobileNet_V2_Weights

weights = MobileNet_V2_Weights.DEFAULT
model = mobilenet_v2(weights=weights)
model.eval()
print("---")
```

```python
# Print parameter names and shapes
for name, param in model.named_parameters():
    print(f"{name:55} | Shape: {list(param.shape)}")
```

```python
# Print parameter names and shapes
for name, param in model.named_parameters():
    if name == "classifier.1.weight":
        weight_matrix = param.data.cpu().numpy()
```

```python
import numpy as np
from utils import generate_plot

array = weight_matrix.flatten()
np.random.shuffle(array)
shuffled = array.reshape(weight_matrix.shape)
        
generate_plot(shuffled)
```
